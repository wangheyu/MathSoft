# 钱学森先生的考题

从地球上发射一枚火箭，绕过太阳，再返回到地球上来，请列出方程、求出解？

## 简化为三体轨道问题

设太阳、地球、火箭的位置分别为
$$
\mathbf r_S,\quad \mathbf r_E,\quad \mathbf r
$$
火箭质量为 $m(t)$，太阳和地球的质量记为 $M_S$ 和 $M_E$。由万有引力公式，地球和太阳对火箭的引力分别为：
$$
g_S = G\frac{M_S m(t)}{\|\mathbf r - \mathbf r_S\|^2}\frac{\mathbf r - \mathbf r_S}{\|\mathbf r - \mathbf r_S\|}
$$
和
$$
g_E = G\frac{M_E m(t)}{\|\mathbf r - \mathbf r_E\|^2}\frac{\mathbf r - \mathbf r_E}{\|\mathbf r - \mathbf r_E\|}.
$$
因此火箭的引力加速度为
$$
\ddot{\mathbf r} m(t) = -(g_S + g_E)
$$
因此有
$$
\ddot{\mathbf r} = -\mu_S \frac{\mathbf r-\mathbf r_S}{\|\mathbf r-\mathbf r_S\|^3}
-\mu_E \frac{\mathbf r-\mathbf r_E}{\|\mathbf r-\mathbf r_E\|^3}.
$$
其中，$\mu_S = GM_S$, $\mu_E = GM_E$, $G$ 是万有引力常数。此外，地球绕太阳的轨道满足方程：
$$
\ddot{\mathbf r}_E = 
-\mu_S \frac{\mathbf r_E-\mathbf r_S}{\|\mathbf r_E-\mathbf r_S\|^3}.
$$

这个可以看作是火箭轨道的二阶微分方程。如果以太阳为圆点，则有 $\mathbf r_S = \mathbf 0$, 且：
$$
\left\{
\begin{align}
\ddot{\mathbf r} &= 
-\mu_S \frac{\mathbf r}{\|\mathbf r\|^3}
-\mu_E \frac{\mathbf r-\mathbf r_E}{\|\mathbf r-\mathbf r_E\|^3},\\
\ddot{\mathbf r}_E &=
-\mu_S \frac{\mathbf r_E}{\|\mathbf r_E\|^3}.
\end{align}
\right.
$$

## 火箭推力方程

假设火箭在真空中飞行（无阻力），并不受外力。经过 $dt$ 时间后，火箭的质量变化为 $m + dm$, 速度为 $v + dv$. 设火箭喷射燃料的速度为 $u$, 方向向后，则在整体惯性系中，燃料喷射速度为 $v - u$. 根据动量守恒，$t$ 时刻动量等于 $t + dt$ 时刻动量，因此：
$$
mv = (m + dm)(v + dv) + (−dm)(v − u),
$$
展开，并忽略二阶小量，有
$$
mv = mv + mdv + vdm + (−dm)v + (−dm)(−u),
$$
化简，得：
$$
mdv=−udm.
$$
这里 $dm < 0$ （火箭在喷射，损失质量），故 $dv > 0$, 即火箭在加速。将此式改写为：
$$
dv = -u\frac{dm}{m},
$$
并假设燃料喷射速度 $u$ 是常数，初始火箭质量是 $m_0$, 终时质量是 $m_f$，初始火箭速度是 $v_0$, 终时火箭速度是 $v_f$，对上式两边一起积分，有：
$$
\int_{v_0}^{v_f} dv = -u \int_{m_0}^{m_f} \frac{dm}{m},
$$
即
$$
v_f - v_0 = u \log \frac{m_0}{m_f}.
$$
此即齐奥尔科夫斯基火箭方程 (Tsiolkovsky rocket equations)。

从另一个角度，火箭的推力等于火箭的单位时间的动量变化，等于单位时间火箭喷出燃料得到的动量改变：
$$
T = \frac{dp}{dt} = \dot{m} u.
$$
其中 $p$ 是动量，$T$ 是火箭推力。

火箭推力方程结合三体轨道方程，得到最终形式的简化微分方程组：
$$
\left\{
\begin{align}
\ddot{\mathbf r} &= 
-\mu_S \frac{\mathbf r}{\|\mathbf r\|^3}
-\mu_E \frac{\mathbf r-\mathbf r_E}{\|\mathbf r-\mathbf r_E\|^3} + \frac{T(t)}{m(t)} \mathbf u(t),\\
\ddot{\mathbf r}_E &=
-\mu_S \frac{\mathbf r_E}{\|\mathbf r_E\|^3},\\
\dot{m} &= -\frac{T(t)}{I_{sp} g_0}.
\end{align}
\right.
$$
边界条件是：
$$
\mathbf r(0) = \mathbf r_E(0),
$$
表示火箭从地球出发，而结束的时候，返回地球，因此
$$
\mathbf r(t_f) = \mathbf r_E(t_f).
$$
此外，返回时的速度也应该满足入轨速度
$$
\dot{\mathbf r}(t_f) = \dot{\mathbf r}_E(t_f).
$$
上面 $\mathbf u(t)$ 是一个单位向量，即
$$
\|\mathbf u(t)\| = 1,
$$
代表火箭在 $t$ 时刻喷气的方向。$I_{sp}$ 和 $g_0$ 是航空航天界喜欢使用的火箭比冲量和一个常数，前者代表了火箭的性能，一般化学火箭的比冲量可以认为是一个常数 $350 s$（单位是秒），$g_0 = 9.80665 m/s^2$，注意，$g_0$ 不是重力加速度，它就是这样一个常数。火箭的排气速度被认为是
$$
v_e = I_{sp}\cdot g_0.
$$
以上方程显然是一种简化，比如我们忽略了地球半径，因此火箭的出发点实际是在地心而不是地球表面；以及我们的火箭返回要求过于严厉，必须精确击中地心，且此时相对地球的速度为 $0$. 此外，我们还必须补上一些符合物理的条件，比如：
$$
m(0) = m_0,
$$
即火箭满载质量（$t = 0$ 时质量）；
$$
0 \leq T(t) \leq T_{\max}.
$$
即火箭发动机最小推力是 $0$ （熄火），最大推力有上限（由火箭材料和设计决定，不能无限加速）。

注意上述方程并不封闭，因为我们不知道 $\mathbf u(t)$ 和 $T(t)$ 的具体形式。但是如果我们把钱先生的题目要求理解为“设计一条合理的轨道以及一枚火箭完成这个任务”，那么本质上我们需要自己去寻找一条轨道，构建这两个方程，然后给出其他细节。
